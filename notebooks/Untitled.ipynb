```python
test_result=adfuller(df['Sales'])
```

```python
test_result
```

```python

# H0 (null hypothesis) = Data is NOT Stationary
# p-val<0.05 --> Data is Stationary - Reject NULL hypothesis - NOT SEASONAL
```

```python
from statsmodels.graphics.tsaplots import plot_acf,plot_pacf
```

```python
fig = plt.figure(figsize=(12,8))
ax1 = fig.add_subplot(211)
fig = plot_acf(df['Sales'].iloc[13:],lags=40,ax=ax1)
ax2 = fig.add_subplot(212)
fig = plot_pacf(df['Sales'].iloc[13:],lags=40,ax=ax2)
```

```python
# For non-seasonal data
#p=1, d=1, q=0 or 1
from statsmodels.tsa.arima.model import ARIMA
```

```python
model=ARIMA(df['Sales'],order=(1,1,1))
model_fit=model.fit()
```

```python
model_fit.summary()
```

```python
# df['forecast']=model_fit.predict(start=90,end=103,dynamic=True)
# df[['Sales','forecast']].plot(figsize=(12,8))
```

```python
# Step 5: Make forecasts for the next 12 months
forecast_steps = 12
forecast = model_fit.get_forecast(steps=forecast_steps)
forecast_index = pd.date_range(start=df.index[-1], periods=forecast_steps + 1, freq='MS')[1:]
forecast_series = pd.Series(forecast.predicted_mean, index=forecast_index)
forecast_conf_int = forecast.conf_int()

# Step 6: Visualize the results
plt.figure(figsize=(12, 6))
plt.plot(df, label='Original Time Series')
plt.plot(forecast_series, label='Forecast', color='red')
plt.fill_between(forecast_series.index, forecast_conf_int.iloc[:, 0], forecast_conf_int.iloc[:, 1], color='pink', alpha=0.3)
plt.title('ARIMA Forecast for the Next 12 Months')
plt.xlabel('Date')
plt.ylabel('Value')
plt.legend()
plt.show()
```
