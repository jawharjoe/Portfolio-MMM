```python
import pandas as pd
```

```python
df = pd.read_csv("Data/forecast_data_3years.csv",encoding='unicode_escape',sep=';')
```

```python
df
```

```python
df = df[['mois','Spend','Conversions']]
```

```python
df['Spend'] = df['Spend'].replace({',':'.'}, regex=True)
```

```python
df['Conversions'] = pd.to_numeric(df['Conversions'])
df['Spend'] = pd.to_numeric(df['Spend'])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

# Line plot with mois as x-axis and Spend & Conversion as y-axes
plt.figure(figsize=(10, 6))  # Set the figure size

plt.plot(df['mois'], df['Spend'], label='Spend')
plt.plot(df['mois'], df['Conversions'], label='Conversion')

# Customize the plot
plt.xlabel('Mois')
plt.ylabel('Value')
plt.title('Spend and Conversion by Month')
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability
plt.legend()  # Add a legend

plt.grid(True)  # Add grid lines for better v
```
